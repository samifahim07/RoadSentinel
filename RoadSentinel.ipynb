## Dataset Nan value Handling

```python
df.isna().sum()
```

Their are many nan present in this dataset 

## Lets`s Handle the Nan 

```python
for column in df.columns:
    
    if df[column].dtype == 'object':
        df[column] = df[column].fillna(df[column].mode()[0])
        
    else:
        df[column] = df[column].fillna(df[column].mean())
```

# After Handling the Nan lets check the missing values again for ensure it 

```python
df.isna().sum()
```

Successfully Handle the missing values in this present dataset
